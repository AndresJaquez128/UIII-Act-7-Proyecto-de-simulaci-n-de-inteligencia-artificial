```python
print("======================================================")
print("      IA PARA FOTÓGRAFO Y CLIENTE")
print("======================================================")

nombre = input("Ingrese el nombre del fotógrafo: ")

while True:

    print("\n=================== MENÚ ===================")
    print("1. Analizar fotógrafo")
    print("2. Analizar cliente")
    print("3. Simular sesión fotográfica")
    print("4. Salir")
    print("============================================")

    opcion = input("Seleccione una opción: ")

    # ==================================================
    # OPCIÓN 1
    # ==================================================

    if opcion == "1":

        print("\n========== ANÁLISIS DEL FOTÓGRAFO ==========")

        experiencia = int(input("Ingrese años de experiencia: "))
        eventos = int(input("Ingrese cantidad de sesiones realizadas: "))

        if experiencia >= 5 and eventos >= 50:

            print("\nLa IA detectó un fotógrafo profesional.")
            print("Recomendaciones:")
            print("- Puede trabajar en eventos grandes")
            print("- Puede usar modo manual")
            print("- Nivel avanzado de fotografía")

        elif experiencia >= 2:

            print("\nLa IA detectó un fotógrafo intermedio.")
            print("Recomendaciones:")
            print("- Mejorar manejo de iluminación")
            print("- Practicar fotografía nocturna")

        else:

            print("\nLa IA detectó un fotógrafo principiante.")
            print("Recomendaciones:")
            print("- Practicar retratos")
            print("- Aprender composición")
            print("- Usar configuraciones automáticas")

        camara = input("\n¿La cámara es profesional? (si/no): ")

        if camara == "si":

            print("La IA recomienda usar lentes avanzados.")

        else:

            print("La IA recomienda comenzar con luz natural.")

    # ==================================================
    # OPCIÓN 2
    # ==================================================

    elif opcion == "2":

        print("\n========== ANÁLISIS DEL CLIENTE ==========")

        tipo = input("¿Qué tipo de sesión desea? (retrato/evento/paisaje): ")

        presupuesto = int(input("Ingrese presupuesto del cliente: "))

        if tipo == "retrato":

            print("\nLa IA detectó sesión de retrato.")

            if presupuesto >= 5000:

                print("Recomendación:")
                print("- Sesión premium")
                print("- Fondo profesional")
                print("- Edición avanzada")

            else:

                print("Recomendación:")
                print("- Sesión básica")
                print("- Luz natural")
                print("- Edición sencilla")

        elif tipo == "evento":

            print("\nLa IA detectó sesión de evento.")

            if presupuesto >= 10000:

                print("Recomendación:")
                print("- Cobertura completa")
                print("- Uso de flash profesional")
                print("- Fotografía y video")

            else:

                print("Recomendación:")
                print("- Cobertura básica")
                print("- Fotografías esenciales")

        elif tipo == "paisaje":

            print("\nLa IA detectó fotografía de paisaje.")

            print("Recomendaciones:")
            print("- Aprovechar luz natural")
            print("- Buscar lugares abiertos")
            print("- Usar gran angular")

        else:

            print("\nTipo de sesión incorrecto")

    # ==================================================
    # OPCIÓN 3
    # ==================================================

    elif opcion == "3":

        print("\n========== SIMULADOR DE SESIÓN ==========")

        luz = input("¿La iluminación es baja o alta?: ")
        lugar = input("¿La sesión es interior o exterior?: ")

        if luz == "baja" and lugar == "interior":

            print("\nLa IA recomienda:")
            print("- Usar iluminación artificial")
            print("- Incrementar ISO")
            print("- Usar trípode")

        elif luz == "alta" and lugar == "exterior":

            print("\nLa IA recomienda:")
            print("- Aprovechar luz natural")
            print("- Reducir brillo")
            print("- Usar filtro UV")

        elif luz == "alta" and lugar == "interior":

            print("\nLa IA recomienda:")
            print("- Ajustar exposición")
            print("- Reducir flash")

        elif luz == "baja" and lugar == "exterior":

            print("\nLa IA recomienda:")
            print("- Activar modo nocturno")
            print("- Usar velocidad lenta")

        else:

            print("\nDatos incorrectos")

    # ==================================================
    # SALIR
    # ==================================================

    elif opcion == "4":

        print("\nGracias por usar la IA Fotográfica,", nombre)
        print("Programa finalizado.")
        break

    # ==================================================
    # ERROR
    # ==================================================

    else:

        print("\nOpción incorrecta")
print("Programa hecho por Jaquez Andres NL = 31")
```
