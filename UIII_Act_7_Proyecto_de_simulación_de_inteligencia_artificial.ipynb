```python
print("======================================================")
print("         IA PARA FOTOGRAFÍA INTELIGENTE")
print("======================================================")

nombre = input("Ingrese su nombre: ")

while True:

    print("\n=================== MENÚ ===================")
    print("1. Analizar iluminación")
    print("2. Analizar tipo de fotografía")
    print("3. Consejos para cámara")
    print("4. Salir")
    print("============================================")

    opcion = input("Seleccione una opción: ")

    # ==================================================
    # OPCIÓN 1
    # ==================================================

    if opcion == "1":

        print("\n======= ANÁLISIS DE ILUMINACIÓN =======")

        print("1. Iluminación baja")
        print("2. Iluminación media")
        print("3. Iluminación alta")

        luz = input("Seleccione una opción: ")

        if luz == "1":

            print("\nLa IA detectó poca iluminación.")
            print("Recomendaciones:")
            print("- Usa trípode")
            print("- Incrementa el ISO")
            print("- Activa modo nocturno")

        elif luz == "2":

            print("\nLa IA detectó iluminación equilibrada.")
            print("Recomendaciones:")
            print("- Mantén brillo automático")
            print("- Usa enfoque automático")
            print("- Ajusta exposición")

        elif luz == "3":

            print("\nLa IA detectó mucha iluminación.")
            print("Recomendaciones:")
            print("- Reduce brillo")
            print("- Usa filtro UV")
            print("- Evita sobreexposición")

        else:

            print("\nOpción incorrecta")

    # ==================================================
    # OPCIÓN 2
    # ==================================================

    elif opcion == "2":

        print("\n======= TIPO DE FOTOGRAFÍA =======")

        print("1. Retrato")
        print("2. Paisaje")
        print("3. Evento")

        foto = input("Seleccione una opción: ")

        if foto == "1":

            print("\nLa IA detectó fotografía de retrato.")
            print("Consejos:")
            print("- Enfoca los ojos")
            print("- Usa fondo desenfocado")
            print("- Mantén buena iluminación")

        elif foto == "2":

            print("\nLa IA detectó fotografía de paisaje.")
            print("Consejos:")
            print("- Usa gran angular")
            print("- Mantén horizonte recto")
            print("- Aprovecha luz natural")

        elif foto == "3":

            print("\nLa IA detectó fotografía de evento.")
            print("Consejos:")
            print("- Usa flash automático")
            print("- Captura momentos rápidos")
            print("- Mantén velocidad alta")

        else:

            print("\nOpción incorrecta")

    # ==================================================
    # OPCIÓN 3
    # ==================================================

    elif opcion == "3":

        print("\n======= CONSEJOS PARA CÁMARA =======")

        bateria = int(input("Ingrese porcentaje de batería: "))

        if bateria >= 80:

            print("\nLa IA detectó batería excelente.")
            print("Puedes realizar sesiones largas.")

        elif bateria >= 40:

            print("\nLa IA detectó batería media.")
            print("Se recomienda llevar cargador.")

        else:

            print("\nLa IA detectó batería baja.")
            print("Recomendaciones:")
            print("- Cargar cámara")
            print("- Reducir brillo")
            print("- Evitar uso de flash")

        memoria = int(input("\nIngrese espacio disponible en GB: "))

        if memoria >= 50:

            print("La IA detectó suficiente almacenamiento.")

        elif memoria >= 10:

            print("La IA recomienda liberar espacio.")

        else:

            print("La IA detectó poco almacenamiento.")
            print("Podrías perder fotografías.")

    # ==================================================
    # SALIR
    # ==================================================

    elif opcion == "4":

        print("\nGracias por utilizar la IA Fotográfica,", nombre)
        print("Programa finalizado.")
        break

    # ==================================================
    # ERROR
    # ==================================================

    else:

        print("\nOpción incorrecta")
print("Programa hecho por Jaquez Andres NL = 31")
```
